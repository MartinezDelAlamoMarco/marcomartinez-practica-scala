## Ejercicio 11

Se pide construir una lista utilizando `Nil` y el operador `::` con los valores:
- Ana
- Luis
- Marta

Y luego otra lista con:
- Pedro
- Sofia

Y se pide concatenar ambas listas con `:::`.

Por último, se pide mostrar:
- Primera lista.
- Segunda lista.
- Lista resultante.

````scala
val lista1: List[String] = "Ana" :: "Luis" :: "Marta" :: Nil
val lista2: List[String] = "Pedro" :: "Sofia" :: Nil

println("Antes de concatenar:")
println("lista1: " + lista1)
println("lista2: " + lista2)

val listaConcatenada: List[String] = lista1 ::: lista2

println()
println("Después de concatenar:")
println("lista1: " + lista1)
println("lista2: " + lista2)
println("listaConcatenada: " + listaConcatenada)
````

### Explicación
- Nil es una lista vacía, y el operador `::` añade un elemento al principio de una lista. Por eso toda lista construida con `::` debe terminar en Nil.
    - "Ana" :: "Luis" :: "Marta" :: Nil se lee de derecha a izquierda: primero se añade Marta a Nil, después Luis y por último Ana.
- El operador `:::` une 2 listas y devuelve una lista nueva (listaConcatenada).
- Como se puede observar en la salida, lista1 y lista2 tienen el mismo contenido antes y después de la concatenación, ya que las listas (List) en Scala son inmutables. El operador `:::` no modifica las listas originales, sino que crea una nueva.

## Ejercicio 12

Con las siguientes puntuaciones:
- val handA = 18
- val handB = 21
- val handC = 25

Se pide crear expresiones que respondan a las siguientes preguntas:
1. ¿Es `handA` mayor que `handB`?
2. ¿Es `handB` igual a 21?
3. ¿Es `handC` diferente de 21?
4. ¿Son `handA` y `handB` menores o iguales a 21?
5. ¿Alguna de las manos `handA` o `handC` supera 21?
6. ¿No se ha pasado `handB` de 21?

Utilizando:
- \>
- <
- \>=
- <=
- ==
- !=
- &&
- ||
- !

````scala
val handA = 18
val handB = 21
val handC = 25

handA > handB
handB == 21
handC != 21
(handA <= 21) && (handB <= 21)
(handA > 21) || (handC > 21)
!(handB > 21)
````

### Explicación

| Expresión | Resultado | Explicación |
| --- | --- | --- |
| handA > handB | false | 18 no es mayor que 21 |
| handB == 21 | true | handB vale exactamente 21 |
| handC != 21 | true | 25 es distinto de 21 |
| (handA <= 21) && (handB <= 21) | true | 18 <= 21 es true y 21 <= 21 es true. Con && deben cumplirse ambas condiciones, y se cumplen |
| (handA > 21) \|\| (handC > 21) | true | 18 > 21 es false, pero 25 > 21 es true. Con \|\| basta con que se cumpla una de las condiciones |
| !(handB > 21) | true | handB > 21 es false (21 no es mayor que 21), y el operador ! lo invierte a true |

## Ejercicio 13

Se pide utilizar:
```val manos = Array(17, 24, 21, 26, 18)```
y la función `bust`.

Se piden 2 versiones:
- Versión A: Bucle while
- Versión B: Utilizando foreach

### Versión A

````scala
val manos = Array(17, 24, 21, 26, 18)

var i = 0
while (i < manos.length) {
    println(manos(i) + ": " + bust(manos(i)))
    i += 1
}
````

### Versión B

````scala
val manos = Array(17, 24, 21, 26, 18)

manos.foreach(mano => println(mano + ": " + bust(mano)))
````

### Explicación
- La versión A (while) es la que necesita un contador, ya que se ejecutará infinitamente hasta que se deje de cumplir la condición, y si los elementos de la condición no varían, nunca se dejará de cumplir, derivando en un bucle infinito.
- La versión A (while) es la que necesita una variable `var` para recorrer la colección, ya que el contador necesita una variable que se vaya incrementando (o reduciendo) para que se deje de cumplir cuando se recorra el array entero, puesto que he decidido que se repita tantas veces como elementos haya en el array, utilizando la propiedad "length".
- La versión B (foreach) es la que se aproxima más al estilo funcional, ya que no utiliza contador ni variables `var`: se le pasa una función que se aplica a cada elemento, y es el propio foreach quien se encarga de recorrer el array.

## Ejercicio 14

**Se pide lo siguiente:**

Analiza el siguiente planteamiento:
```
var total = 0
def sumarAlTotal(valor: Int) = {
  total = total + valor
}
```

Realiza las siguientes tareas:
1. Ejecuta la función varias veces.
2. Muestra el valor de total después de cada llamada.
3. Explica por qué la función modifica una variable situada fuera de su ámbito local.
3. Identifica el efecto secundario.

Después crea una segunda versión:
```
def sumar(a: Int, b: Int): Int = {
  ...
}
```
que reciba valores y devuelva un nuevo resultado sin modificar variables externas.

### Primera versión

````scala
var total = 0

def sumarAlTotal(valor: Int) = {
  total = total + valor
}

sumarAlTotal(1)
println(s"Total después de la primera llamada: ${total}")
sumarAlTotal(-5)
println(s"Total después de la segunda llamada: ${total}")
sumarAlTotal(3)
println(s"Total después de la tercera llamada: ${total}")
sumarAlTotal(7)
println(s"Total después de la cuarta llamada: ${total}")
sumarAlTotal(4)
println(s"Total después de la quinta llamada: ${total}")
````

### Segunda versión

````scala
def sumar(a: Int, b: Int): Int = {
  a + b
}

println(s"Suma: ${sumar(1, 4)}")
println(s"Suma: ${sumar(-5, 6)}")
println(s"Suma: ${sumar(3, 10)}")
println(s"Suma: ${sumar(7, 1)}")
println(s"Suma: ${sumar(4, -4)}")
````

### Explicación
Por qué la función modifica una variable fuera de su ámbito local:
- La variable total no está declarada dentro de la función, sino fuera, en un ámbito superior. Scala permite que una función acceda a las variables visibles en el lugar donde ha sido definida, y como total es una var (mutable), la función no solo puede leerla, sino también reasignarla.

Efecto secundario:
- El efecto secundario es la modificación de la variable total. La función no devuelve ningún valor, lo único que hace es cambiar el estado de una variable externa.
- La consecuencia es que el resultado depende del orden y del número de llamadas: sumarAlTotal(3) no produce siempre lo mismo, ya que depende de cuánto valga total en ese momento. En cambio, sumar(3, 10) devuelve siempre 13, sin importar cuándo se ejecute.

### Comparación
| Característica | sumarAlTotal | sumar |
| --- | --- | --- |
| Modifica datos externos | Sí | No |
| Devuelve un resultado calculado | No | Sí |
| Utiliza efecto secundario | Sí | No |
| Estilo predominante | Imperativo | Funcional |

## Ejercicio 15 - Programa integrado: torneo de Twenty-One

Desarrolla un pequeño programa que combine los principales conceptos trabajados en esta parte.

Dispones de los siguientes jugadores:

```scala
val jugadores = List("Alex", "Chen", "Marta", "Sindhu")
```

y de sus puntuaciones:

```scala
val manos = Array(18, 24, 21, 20)
```

El programa deberá:

1. Mantener los nombres de los jugadores en una `List`.
2. Mantener las puntuaciones en un `Array[Int]`.
3. Utilizar la función `bust` para determinar qué manos superan 21.
4. Recorrer todas las puntuaciones y mostrar si cada mano se ha pasado o no.
5. Determinar cuál es la mejor puntuación válida.
6. Utilizar `if`, `else if` y `else` cuando sea necesario.
7. Utilizar operadores relacionales o lógicos.
8. Utilizar al menos una función definida por ti.
9. Utilizar al menos una estructura `while` o `foreach`.
10. Evitar modificar las listas originales.

### Ampliación dentro del mismo ejercicio

Añade una segunda ronda:

```scala
val manosRonda2 = Array(22, 19, 20, 21)
```

Determina nuevamente qué jugadores continúan con una puntuación válida y cuál es la mejor puntuación de la ronda.

### Informe final del ejercicio

Añade una celda Markdown explicando:

- Qué partes de tu solución son mutables.
- Qué partes son inmutables.
- Dónde utilizas funciones.
- Dónde utilizas estructuras de control.
- Qué parte de tu solución consideras más cercana al estilo funcional.
- Qué parte consideras más cercana al estilo imperativo.

````scala
val jugadores = List("Alex", "Chen", "Marta", "Sindhu")
val manos = Array(18, 24, 21, 20)

def compruebaPuntuaciones(manos: Array[Int]) = {
    println("Qué manos superan 21:")
    var i = 0
    while(i < jugadores.length) {
        print(s"- La mano de ${jugadores(i)}: ")
        println(
            if (bust(manos(i))) "Sí"
            else "No"
            )
        i += 1
    }
}

def manoMayor(manos: Array[Int]) = {
    var indexManoMayor = -1
    var j = 0
    
    while(j < manos.length) {
        if (!bust(manos(j))) {
            if (indexManoMayor == -1 || manos(j) > manos(indexManoMayor)) {
                indexManoMayor = j
            }
        }
        j += 1
    }

    if (indexManoMayor == -1) println("Ningún jugador tiene una puntuación válida en esta ronda")
    else println(s"La mejor mano válida es la de ${jugadores(indexManoMayor)}: ${manos(indexManoMayor)}")
}

println("RONDA 1")
compruebaPuntuaciones(manos)
println()
manoMayor(manos)
println()

val manosRonda2 = Array(22, 19, 20, 21)

println("RONDA 2")
compruebaPuntuaciones(manosRonda2)
println()
manoMayor(manosRonda2)
````

### Explicación
Partes mutables:
- Las variables `i`, `j` e `indexManoMayor`, declaradas como `var` dentro de las funciones. Las dos primeras son contadores de los bucles, y la tercera va guardando el índice de la mejor mano válida encontrada hasta el momento.
- Los Array (`manos` y `manosRonda2`) son mutables por naturaleza, ya que permiten modificar sus elementos, aunque en mi solución únicamente los leo.

Partes inmutables:
- La `List` de jugadores, que además de estar declarada como `val`, es inmutable por naturaleza.
- Todas las referencias declaradas con `val`, que no se reasignan en ningún momento.
- La función bust, que no modifica nada: recibe una puntuación y devuelve un Boolean.

Dónde utilizo funciones:
- bust, definida en el ejercicio 4, para determinar si una mano supera 21.
- compruebaPuntuaciones, que recibe las manos de una ronda y muestra si cada jugador se ha pasado o no.
- manoMayor, que recibe las manos de una ronda y muestra la mejor puntuación válida.
- Las dos últimas reciben las manos por parámetro, por lo que no dependen de ninguna variable externa y sirven para cualquier ronda. Por eso he podido reutilizarlas en la ronda 2 sin modificar nada.

Dónde utilizo estructuras de control:
- Dos bucles `while`, uno en cada función, para recorrer las colecciones mediante un índice.
- `if` / `else` en compruebaPuntuaciones, para mostrar si la mano se ha pasado o no.
- `if` / `else` en manoMayor, tanto para decidir si una mano es la nueva mejor como para contemplar el caso en el que ningún jugador tenga una puntuación válida.
- Operadores relacionales y lógicos: `<`, `>`, `==`, `!` y `||`.

Parte más cercana al estilo funcional:
- La función bust, ya que es una función pura: recibe un valor, devuelve un resultado y no modifica nada externo. También la `List` de jugadores, al ser inmutable.

Parte más cercana al estilo imperativo:
- Los bucles `while` con sus contadores `var`, y la variable `indexManoMayor`, que se va reasignando conforme se recorre el array. Esta parte se podría escribir de forma funcional con `foreach` o `filter` y `max`, sin contadores ni variables mutables.
