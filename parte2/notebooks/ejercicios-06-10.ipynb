## Funciones auxiliares

````scala
def bust(puntuacion: Int): Boolean = {
    puntuacion > 21
}

def maxHand(hand1: Int, hand2: Int): Int = {
    if (hand1 > hand2) hand1
    else hand2
}
````

## Ejercicio 6

Se pide utilizar la función bust del ejercicio 4 y definir una función llamada ganador que reciba 2 manos y:
1. Si ambas manos superan 21, devuelve 0.
2. Si solamente handA supera 21, devuelve handB.
3. Si solamente handB supera 21, devuelve handA.
4. Si ninguna supera 21, devuelve la mano con mayor puntuación.

El enunciado no lo especifica, pero voy a utilizar también la función maxHand.

````scala
def ganador(handA: Int, handB: Int): Int = {
    if (bust(handA) && bust(handB)) 0
    else if (bust(handA) && !bust(handB)) handB
    else if (!bust(handA) && bust(handB)) handA
    else maxHand(handA, handB)
}

ganador(26, 20)
ganador(18, 22)
ganador(24, 25)
ganador(17, 19)
ganador(21, 20)
````

### Explicación
| Prueba | Se cumple | Parte de código |
| --- | --- | --- |
| 1 (26, 20) | Condición 2 | else if (bust(handA) && !bust(handB)) handB |
| 2 (18, 22) | Condición 3 | else if (!bust(handA) && bust(handB)) handA |
| 3 (24, 25) | Condición 1 | if (bust(handA) && bust(handB)) 0 |
| 4 (17, 19) | Condición 4 | else maxHand(handA, handB) |
| 5 (21, 20) | Condición 4 | else maxHand(handA, handB) |

## Ejercicio 7

Se pide declarar un array y realizar las siguientes operaciones:
1. Muestra el array original.
2. Sustituye "Alex" por "Sindhu".
3. Muestra nuevamente el array.
4. Intenta asignar el número 500 a una de sus posiciones.

````scala
val jugadores = Array("Alex", "Chen", "Marta")

println("Array antes de modificar: " +jugadores.mkString(", "))
jugadores(0) = "Sindhu"

println("Array después de modificar: " +jugadores.mkString(", "))

//jugadores(2) = 500 //Comento la línea para que no de error. Si se descomenta dará error
````

````scala
// DEMOSTRACIÓN DEL ERROR QUE DEBERÍA SALIR
jugadores(2) = 500
````

### Explicación
- Los elementos del array se pueden cambiar aunque la variable se haya declarado con val, porque lo que es inmutable es la referencia al array, los elementos sí que son mutables.
- Scala no permite introducir el valor `500` en un Array\[String\] porque los elementos de dentro del array deben ser de tipo texto y el valor `500` es de tipo numérico, si se quisiera incluir el número 500, se debería poner entre comillas (`"500"`)
- Reasignar la variable significa cambiar el valor la propia variable llamada `jugadores`, es decir, que por ejemplo después de declarar ```val jugadores = Array("Alex", "Chen", "Marta")```, se intenta reasignar a otro valor: ```jugadores = Array("Ana", "Luis")``` (no permitido, ya que es de una variable "val", y no "var"). Modificar un elemento del array significa que dentro del propio array se cambia el valor de una de las variables, como por ejemplo ```jugadores(0) = "Sindhu"``` (sí permitido, ya que las variables de dentro de un array sí que son mutables).

## Ejercicio 9

Se pide utilizar el siguiente array: 
- val manos = Array(17, 24, 21, 19, 26)

Después, se pide recorrerlo con un bucle while y para cada posición mostrar:
- El valor de la mano.
- El resultado de aplicar la función `bust`.

Se pide utilizar:
- var i = 0
- manos.length

````scala
var i = 0
val manos = Array(17, 24, 21, 19, 26)

while (i < manos.length) {
    println("Valor de la mano: " + manos(i))
    println("Resultado de la función bust: " + bust(manos(i)))
    i += 1
}
````

### Explicación
- El contador debe ser mutable porque el contador debe incrementar para que el bucle se ejecute el número de veces que se desee.
- Si no incrementase `i`, se volvería un bucle infinito, el cual repetiría la función con el primer valor infinitamente. Esto dará lugar a problemas de rendimiento.
- Elementos que corresponden al estilo imperativo:
    - var i = 0 --> variable mutable.
    - while --> estructura de control imperativa.
    - i += 1 --> modificación del estado de una variable.
    - Acceso mediante manos(i) --> recorrido controlado manualmente mediante un índice.

## Ejercicio 10

Se pide crear una lista con nombres de jugadores, para posteriormente:
1. Muestra la lista original.
2. Crea una nueva lista añadiendo `"Sindhu"` al principio mediante `::`.
3. Muestra ambas listas.
4. Comprueba que la lista original no ha cambiado.
5. Muestra:
    - Su longitud.
    - La lista invertida mediante `reverse`.

````scala
val jugadores = List("Alex", "Chen", "Marta")

println("Lista original: " + jugadores.mkString(", "))

val nuevosJugadores: List[String] = "Sindhu" :: jugadores

println("Lista original: " + jugadores.mkString(", "))
println("Nueva lista: " + nuevosJugadores.mkString(", "))

// Efectivamente la lista original no ha cambiado en absoluto.

println("Longitud lista original: " + jugadores.length)
println("Lista original invertida: " + jugadores.reverse.mkString(", "))

println("Longitud lista nueva: " + nuevosJugadores.length)
println("Lista nueva invertida: " + nuevosJugadores.reverse.mkString(", "))
````

EL operador `::` produce una nueva lista en lugar de modificar la lista original porque el tipo de dato `List` es inmutable, por lo que para añadir un nuevo elemento se deberá crear una lista, o bien reasignar la variable (en caso de que la variable sea `var` y lo permita).
